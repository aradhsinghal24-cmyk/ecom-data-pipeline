```python
# Databricks notebook source
# MAGIC %md
# MAGIC # 99 - Cleanup
# MAGIC Drops all pipeline tables. Optional — run only when you want a fresh start.

dbutils.widgets.text("catalog", "delta_catalog")
dbutils.widgets.text("schema",  "delta_demo")
CATALOG = dbutils.widgets.get("catalog")
SCHEMA  = dbutils.widgets.get("schema")

# COMMAND ----------

for tbl in ["landing_orders", "landing_customers", "bronze_orders",
            "silver_orders", "silver_customers", "gold_sales_summary"]:
    spark.sql(f"DROP TABLE IF EXISTS {CATALOG}.{SCHEMA}.{tbl}")
    print(f"Dropped: {tbl}")

print("All tables dropped.")
```
